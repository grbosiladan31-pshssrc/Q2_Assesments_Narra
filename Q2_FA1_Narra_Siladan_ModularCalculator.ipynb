```python
Python
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operation = input("Enter operation: ")


def add_numbers (num1, num2):
    sum = num1 + num2
    return sum

def subtract_numbers (num1, num2):
    difference = num1 - num2
    return difference

def multiply_numbers (num1, num2):
    product = num1 * num2
    return product

def divide_numbers (num1, num2):
    quotient = num1 / num2
    return quotient


if operation == "addition" or "Adittion":
    print(add_numbers,(num1 , num2))

elif operation == "subtraction" or "Substraction":
    print(subtract_numbers,(num1 , num1))

elif operation == "multiplication" or "Multiplication":
    print(multiply_numbers,(num1, num2))

elif operation == "division" or "Division":
    print(divide_numbers)
```
